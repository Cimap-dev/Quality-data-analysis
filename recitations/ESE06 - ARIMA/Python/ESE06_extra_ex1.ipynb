# EXTRA EXERCISE 1

The number of hours required to complete an assembly operation of a vehicle is monitored for each unit. 
The data is reported in the file ‘hours.csv’. 

1. Inspect the dataset and verify the assumptions. 
2. Suggest a suitable model for the data. 



```python
# Import the necessary libraries
import statsmodels.api as sm
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats
import seaborn as sns
import qdatoolkit as qda

# Import the dataset
data = pd.read_csv('../../Data/hours.csv')

# Inspect the dataset
data.head()
```

## Point 1
Inspect the dataset and verify the assumptions. 



```python
# Plot the data first
plt.plot(data['Hours'], 'o-')
plt.xlabel('Index')
plt.ylabel('PH measuremet')
plt.title('Time series plot of pH measurements')
plt.grid()
plt.show()
```

It looks like the data are either following a trend or a meandering pattern. 


```python
# Normality
_ = qda.Assumptions(data['Hours']).normality()
```

p-value < 0.01, hence we **reject** the normality assumption


```python
# Randomness 
_ = qda.Assumptions(data['Hours']).independence()
```

The sample ACF follows a linear decay, so non-stationarity needs to be addressed.

We reject the randomness.

Let’s try the differencing operator:

```python
#calculate the lag1 from data
data['diff1'] = data['Hours'].diff(1)

plt.plot(data['diff1'], 'o-')
plt.xlabel('Index')
plt.ylabel('DIFF 1')
plt.title('Time series plot of DIFF 1')
plt.grid()
plt.show()
```

```python
# Normality of diff1
_ = qda.Assumptions(data['diff1']).normality()
```

```python
# Randomness of diff1
_ = qda.Assumptions(data['diff1']).independence()
```

## Point 2

We can fit an **AR(1)** model to this new dataset (equal to ARIMA(1,1,0) on the original dataset). 


```python
# fit model ARIMA with constant term
x = data['diff1'][1:]
model = qda.ARIMA(x, order=(1,0,0), add_constant = True) 

qda.ARIMAsummary(model)
```

All the terms are significant. 
Let’s check if the assumptions on the residuals are met. 


```python
# Normality of residuals
_ = qda.Assumptions(model.resid).normality()
```

```python
# Randomness of residuals
_ = qda.Assumptions(model.resid).independence()
```

The model is adequate.
